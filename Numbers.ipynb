# Numbers in Python

## Signed Numbers:
> Positive numbers don't need a sign; However, to refer to a negative number, one must add `-` before the number.

```python
4
```

```python
-28
```

## Operations on numbers:
> Python can add numbers using `+`, subtract numbers using `-`, multiply numbers using `*` and devide numbers using `/`. The power operator is `**`.

```python
2 + 4
```

```python
8 - 3
```

```python
8 - 20
```

```python
4 * 3
```

```python
18 / 3
# The type of the devision (/)is always float.
```

```python
2 ** 3
```

### There is another two operators:
> To have an integer after devision, instead of `/`, one must use `//`. There is also another useful operator called modulo `%` to calculate the remainder of a devision.

```python
18 // 7
```

```python
18 / 7
```

```python
18 % 7
```

> To define if a number is even we can use `%`, if the remainder is 0, it is even, if it returns 1, it's an odd number.

```python
5 % 2
```

```python
67 % 2
```

```python
88 % 2
```

## Operatores Precedence:
> The priority is the `()`, then `**`, then `*, /`, eventually `+, -`. For a more readable code, it is recommended to use enough parentheses.

```python
2 ** 3 * 4 + 8 / 2
```

```python
((2 ** 3) * 4) + (8 / 2)
```

> The underscore `_` sign can be used for readability in large numbers.

```python
1000000000
```

```python
1_000_000_000
```

```python
24567288
```

```python
24_567_288
```
