## Indexing in Strings
> Indexing here is similar to lists, etc.

```python
message[0]
```

```python
message [-1] # Call the last character
```

```python
message [::]
```

```python
message [1:5]
```

```python
message [0:5:2]  # Start from the first element to the 5th element with 2 steps
```

### Length of a string

```python
len(message)
```

### The most complete format of indexing:
> message [start:end:step]

```python
ns = '0123456789'
ns [2:9:3]
```

```python
ns [1:-1]
```

### Reversing
> `[::]` mean the whole stirng, `-1` for the step means that start from the end and reversely go through the string's charachters

```python
ns [::-1]  
```

### f string

```python
age = 22
height = 184
weight = 80.298710434
```

```python
print (f"Hello, {name}, Happy to see you. I think you are {age} years old, and {height} tall. Also, I noticed that you are {weight:1.4f} kiolograms.")
```

> Precision: `{weight:1.4f}`; `1`: number of free spaces, `4`: number of numbers after `.`, and `f` stands for float.
