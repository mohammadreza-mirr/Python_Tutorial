# Dictionaries in Python

> Dictionaries are a combination of key and value in `{}`

```python
fruits_price = {'Bannana': 2, 'Apple': 5, 'Melon': 12, 'Berry': 8}
```

> We can call a key in the dictionary, with indexing:

```python
fruits_price ['Bannana']
```

```python
grades = {
    'John': 
    {
        'Math': 3, 'Programming':3.5, 'Physics': 2.1
    },
    'Mike':
    {
        'Math': 2, 'Programming':4, 'Physics': 1.4
    },
    'Sarah':
    {
        'Math': 3.8, 'Programming':4, 'Physics': 3.9
    }
}
```

```python
grades ['Sarah']
```

```python
grades ['John']['Math']
```

## Dictionaries Methods

```python
grades.keys()
```

```python
grades.values()
```

> `get` function returns the values of the called key, also we can assign a vlaue (e.g., -1), such that if that key is not included in the dictionary, it returns the assigned value.

```python
grades.get('Donald', -1)
```

```python
grades.items()
```

### Combination of methods

```python
classes = {'A': ['John', 'Mike', 'Joe'], 'B':['Sarah', 'John','Dan']}
```

```python
classes['B'][2].upper()
```

```python
classes ['A'][-1]
```

```python
last_person = classes ['B'].pop()
```

```python
last_person
```
