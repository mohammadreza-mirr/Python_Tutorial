# Variables in Python
> A name we typically use to assign a value to save the value, so that we can call that later in our program.

```python
pizza_price = 28
number = 12
soda_price = 6
```

```python
total_value = (pizza_price * number) + (soda_price * number)
print (total_value)
```

> values in Python are intiated with `a-z` or `A-Z` (Python is letter sensetive, i.e., ther is difference between A and a.) and after the first letter we can use numbers, and underscore (e.g., num1, car_price, Toala_value, etc.)

> Python is Dynamic Type, meaning that we can change the value of a variable during the program.

```python
number = 4
```

```python
number = number + 3
```

```python
number
```

> To increment a certain value of a variable with another value, we can use: ` variable_name =+ value to be incremented`

```python
number += 1
number
```
