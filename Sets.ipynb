# Sets in Python

```python
my_set = set()
```

```python
my_set.add(1)
my_set.add (2)
```

```python
type(my_set)
```

```python
chess = set()
chess.add("Mike")
chess.add("Sarah")
```

```python
pot = set()
pot.add("Sarah")
pot.add("Joe")
```

```python
chess.union(pot)
```

```python
chess.intersection(pot)
```
